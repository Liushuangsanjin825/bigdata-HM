## 1. 数据概览（Data Understanding）


```python
def load_tables(base_path: str) -> dict[str, pl.LazyFrame]:
    base = Path(base_path)
    return {
        "articles": pl.scan_csv(
            base / "articles.csv",
            schema_overrides={
                "article_id": pl.Utf8,
                "product_code": pl.Utf8,
            },
        ),
        "customers": pl.scan_csv(
            base / "customers.csv",
            schema_overrides={
                "customer_id": pl.Utf8,
                "postal_code": pl.Utf8,
            },
        ),
        "transactions": pl.scan_csv(
            base / "transactions_train.csv",
            schema_overrides={
                "t_dat": pl.Utf8,
                "customer_id": pl.Utf8,
                "article_id": pl.Utf8,
            },
        ),
        "submission": pl.scan_csv(
            base / "sample_submission.csv",
            schema_overrides={"customer_id": pl.Utf8},
        ),
    }


def profile_table(lf: pl.LazyFrame, name: str) -> pl.DataFrame:
    rows = lf.select(pl.len().alias("rows")).collect(engine="streaming")["rows"][0]
    schema = lf.collect_schema()
    return pl.DataFrame(
        {
            "table": [name],
            "rows": [int(rows)],
            "columns": [len(schema.names())],
            "column_list": [", ".join(schema.names())],
        }
    )
```

```python
tables = load_tables(str(BASE_PATH))

df_profiles = pl.concat(
    [profile_table(tables[k], k) for k in ["articles", "customers", "transactions", "submission"]],
    how="vertical_relaxed",
)

df_profiles
```

## 2. 质量诊断（Data Quality Audit）


```python
def missing_ratio_report(lf: pl.LazyFrame, dataset: str) -> pl.DataFrame:
    schema = lf.collect_schema()
    exprs = [
        (pl.col(c).null_count() / pl.len()).alias(c)
        for c in schema.names()
    ]
    out = lf.select(exprs).collect(engine="streaming")
    long_df = out.unpivot(variable_name="column", value_name="null_ratio")
    return long_df.with_columns(pl.lit(dataset).alias("dataset")).select(["dataset", "column", "null_ratio"])


df_missing_report = pl.concat(
    [
        missing_ratio_report(tables["customers"], "customers"),
        missing_ratio_report(tables["articles"], "articles"),
        missing_ratio_report(tables["transactions"], "transactions"),
    ],
    how="vertical_relaxed",
).sort(["dataset", "null_ratio"], descending=[False, True])


df_missing_report.head(30)
```

```python
# 1) Duplicate row check on all transaction columns
all_tx_cols = tables["transactions"].collect_schema().names()

df_dup = tables["transactions"].select(
    [
        pl.len().alias("rows_total"),
        pl.struct(all_tx_cols).n_unique().alias("rows_unique"),
    ]
).collect(engine="streaming")

rows_total = int(df_dup["rows_total"][0])
rows_unique = int(df_dup["rows_unique"][0])
dup_rows = rows_total - rows_unique

# 2) Category anomaly check (e.g., NONE vs None)
df_news_levels = (
    tables["customers"]
    .select(pl.col("fashion_news_frequency").fill_null("NULL").alias("fashion_news_frequency"))
    .group_by("fashion_news_frequency")
    .agg(pl.len().alias("cnt"))
    .sort("cnt", descending=True)
    .collect(engine="streaming")
)

# 3) Numeric quantiles (price)
df_price_quantiles = tables["transactions"].select(
    [
        pl.col("price").quantile(0.01).alias("price_q01"),
        pl.col("price").quantile(0.50).alias("price_q50"),
        pl.col("price").quantile(0.95).alias("price_q95"),
        pl.col("price").quantile(0.99).alias("price_q99"),
    ]
).collect(engine="streaming")

# 4) Join coverage (transactions x article metadata)
df_join_cov = (
    tables["transactions"]
    .join(tables["articles"].select(["article_id", "product_group_name"]), on="article_id", how="left")
    .select(
        [
            pl.len().alias("rows"),
            pl.col("product_group_name").null_count().alias("article_meta_missing"),
        ]
    )
    .collect(engine="streaming")
)

# Consolidated quality table for grading checkpoints
quality_rows = [
    {"metric": "transactions_rows_total", "value": str(rows_total)},
    {"metric": "transactions_rows_unique", "value": str(rows_unique)},
    {"metric": "transactions_duplicate_rows", "value": str(dup_rows)},
    {"metric": "price_q01", "value": f"{float(df_price_quantiles['price_q01'][0]):.6f}"},
    {"metric": "price_q50", "value": f"{float(df_price_quantiles['price_q50'][0]):.6f}"},
    {"metric": "price_q95", "value": f"{float(df_price_quantiles['price_q95'][0]):.6f}"},
    {"metric": "price_q99", "value": f"{float(df_price_quantiles['price_q99'][0]):.6f}"},
    {"metric": "article_meta_missing_after_join", "value": str(int(df_join_cov['article_meta_missing'][0]))},
]

df_quality_report = pl.DataFrame(quality_rows)

df_quality_report
```

## 3. 清洗规则制定（Rulebook）


```python
df_clean_rulebook = pl.DataFrame(
    {
        "问题": [
            "customers.FN/Active 缺失率高",
            "club_member_status / fashion_news_frequency 值不规范",
            "age 存在缺失",
            "transactions 存在重复行",
            "price 长尾分布",
            "articles.detail_desc 少量缺失",
        ],
        "清洗策略": [
            "保留原列；新增 FN_flag/Active_flag 与 is_missing_FN/is_missing_Active",
            "统一大写，缺失映射 UNKNOWN，同义值归并",
            "保留 is_missing_age，并并行构造 age_imputed(中位数填补)",
            "默认保留原始交易表，同时构建去重口径供对比",
            "保留原始 price，同时构建 price_winsorized(1%-99%)",
            "detail_desc 填充空字符串，并增加 desc_missing_flag",
        ],
        "代码实现": [
            "with_columns + fill_null + cast + is_null",
            "str.to_uppercase + fill_null + replace",
            "median + fill_null",
            "struct(...).n_unique() / unique()",
            "quantile + clip",
            "fill_null('') + is_null",
        ],
        "潜在风险": [
            "缺失可能有业务含义，不能直接删",
            "归并规则过严可能丢失细粒度信息",
            "年龄填补会引入分布偏移",
            "去重可能误删真实重复购买",
            "裁剪会压缩极端高价信息",
            "文本缺失补空会弱化语义",
        ],
    }
)

df_clean_rulebook
```

```python
def build_clean_customers(lf_customers: pl.LazyFrame) -> pl.LazyFrame:
    age_median = (
        lf_customers
        .select(pl.col("age").drop_nulls().median().alias("age_median"))
        .collect(engine="streaming")["age_median"][0]
    )

    return lf_customers.with_columns(
        [
            pl.col("FN").is_null().cast(pl.Int8).alias("is_missing_FN"),
            pl.col("Active").is_null().cast(pl.Int8).alias("is_missing_Active"),
            pl.col("FN").fill_null(0).cast(pl.Int8).alias("FN_flag"),
            pl.col("Active").fill_null(0).cast(pl.Int8).alias("Active_flag"),
            pl.col("club_member_status").str.to_uppercase().fill_null("UNKNOWN").alias("club_member_status_clean"),
            pl.col("fashion_news_frequency").str.to_uppercase().fill_null("UNKNOWN").alias("fashion_news_frequency_clean"),
            pl.col("age").is_null().cast(pl.Int8).alias("is_missing_age"),
            pl.col("age").fill_null(age_median).alias("age_imputed"),
        ]
    )


def build_clean_transactions(lf_tx: pl.LazyFrame) -> pl.LazyFrame:
    q = (
        lf_tx.select(
            [
                pl.col("price").quantile(0.01).alias("q01"),
                pl.col("price").quantile(0.99).alias("q99"),
            ]
        )
        .collect(engine="streaming")
    )
    q01 = float(q["q01"][0])
    q99 = float(q["q99"][0])

    return lf_tx.with_columns(
        [
            pl.col("t_dat").str.to_date(),
            pl.col("price").clip(q01, q99).alias("price_winsorized"),
        ]
    )
```

```python
lf_customers_clean = build_clean_customers(tables["customers"])
lf_transactions_clean = build_clean_transactions(tables["transactions"])
lf_articles_clean = tables["articles"].with_columns(
    [
        pl.col("detail_desc").is_null().cast(pl.Int8).alias("desc_missing_flag"),
        pl.col("detail_desc").fill_null("").alias("detail_desc_clean"),
    ]
)

print("clean customers sample")
print(lf_customers_clean.select(["customer_id", "FN", "FN_flag", "is_missing_FN", "age", "age_imputed", "fashion_news_frequency", "fashion_news_frequency_clean"]).limit(5).collect(engine="streaming"))

print("\nclean transactions sample")
print(lf_transactions_clean.select(["t_dat", "customer_id", "article_id", "price", "price_winsorized"]).limit(5).collect(engine="streaming"))
```

```python
# 4.4 年龄分布与消费强度（年龄段 x 人均购买次数）
age_bin_expr = (
    pl.when(pl.col("age_imputed") < 18).then(pl.lit("<18"))
    .when(pl.col("age_imputed") < 25).then(pl.lit("18-24"))
    .when(pl.col("age_imputed") < 35).then(pl.lit("25-34"))
    .when(pl.col("age_imputed") < 45).then(pl.lit("35-44"))
    .when(pl.col("age_imputed") < 55).then(pl.lit("45-54"))
    .when(pl.col("age_imputed") < 65).then(pl.lit("55-64"))
    .otherwise(pl.lit("65+"))
)

_df_age_intensity = (
    lf_transactions_clean
    .group_by("customer_id")
    .agg(pl.len().alias("txn_cnt"))
    .join(lf_customers_clean.select(["customer_id", "age_imputed"]), on="customer_id", how="left")
    .with_columns(age_bin_expr.alias("age_bin"))
    .group_by("age_bin")
    .agg([
        pl.len().alias("user_cnt"),
        pl.mean("txn_cnt").alias("avg_txn_per_user"),
    ])
    .with_columns(
        pl.when(pl.col("age_bin") == "<18").then(0)
        .when(pl.col("age_bin") == "18-24").then(1)
        .when(pl.col("age_bin") == "25-34").then(2)
        .when(pl.col("age_bin") == "35-44").then(3)
        .when(pl.col("age_bin") == "45-54").then(4)
        .when(pl.col("age_bin") == "55-64").then(5)
        .otherwise(6)
        .alias("order_idx")
    )
    .sort("order_idx")
    .collect(engine="streaming")
)

fig_age = make_subplots(specs=[[{"secondary_y": True}]])
fig_age.add_trace(go.Bar(x=_df_age_intensity["age_bin"], y=_df_age_intensity["user_cnt"], name="用户数"), secondary_y=False)
fig_age.add_trace(go.Scatter(x=_df_age_intensity["age_bin"], y=_df_age_intensity["avg_txn_per_user"], name="人均购买次数", mode="lines+markers"), secondary_y=True)
fig_age.update_layout(title="年龄段分布与消费强度")
fig_age.update_xaxes(title_text="年龄段")
fig_age.update_yaxes(title_text="用户数", secondary_y=False)
fig_age.update_yaxes(title_text="人均购买次数", secondary_y=True)
fig_age.show()

# [doc-image] 4.4 age intensity
_age_bins = _df_age_intensity["age_bin"].to_list()
_user_cnt = _df_age_intensity["user_cnt"].to_list()
_avg_txn = _df_age_intensity["avg_txn_per_user"].to_list()
_mfig, _ax1 = plt.subplots(figsize=(10.5, 4.8))
_ax1.bar(_age_bins, _user_cnt, color="#4e79a7", alpha=0.85)
_ax1.set_xlabel("age_bin")
_ax1.set_ylabel("user_cnt", color="#4e79a7")
_ax2 = _ax1.twinx()
_ax2.plot(_age_bins, _avg_txn, color="#f28e2b", marker="o", linewidth=1.8)
_ax2.set_ylabel("avg_txn_per_user", color="#f28e2b")
_mfig.suptitle("Age distribution and consumption intensity")
_ax1.grid(axis="y", alpha=0.3)
_save_doc_png(_mfig, "fig_4_4_age_intensity.png")
```

```python
# 4.5 价格分布（原始 / log / winsorized）
_df_price = (
    lf_transactions_clean
    .select(["price", "price_winsorized"])
    .limit(500_000)
    .collect(engine="streaming")
    .with_columns(pl.col("price").log1p().alias("price_log1p"))
)

_df_price_long = _df_price.select(["price", "price_log1p", "price_winsorized"]).unpivot(
    variable_name="metric", value_name="value"
)

fig_price = px.histogram(
    _df_price_long,
    x="value",
    color="metric",
    nbins=80,
    barmode="overlay",
    opacity=0.55,
    title="价格分布对比（采样 50 万行）",
)
fig_price.show()

# [doc-image] 4.5 price distribution comparison
_price = _df_price["price"].to_list()
_price_log = _df_price["price_log1p"].to_list()
_price_w = _df_price["price_winsorized"].to_list()
_mfig, _ax = plt.subplots(figsize=(10.5, 4.8))
_ax.hist(_price, bins=80, alpha=0.35, label="price")
_ax.hist(_price_log, bins=80, alpha=0.35, label="price_log1p")
_ax.hist(_price_w, bins=80, alpha=0.35, label="price_winsorized")
_ax.set_xlabel("value")
_ax.set_ylabel("count")
_ax.set_title("Price distribution comparison")
_ax.legend()
_ax.grid(alpha=0.2)
_save_doc_png(_mfig, "fig_4_5_price_comparison.png")
```

## 5. 初期特征挖掘（Checkpoint核心）


```python
def build_user_features(lf_tx: pl.LazyFrame, cutoff_date: str) -> pl.LazyFrame:
    cutoff_expr = pl.lit(cutoff_date).str.to_date()

    lf_hist = (
        lf_tx
        .filter(pl.col("t_dat") <= cutoff_expr)
        .with_columns([
            (pl.col("t_dat") > (cutoff_expr - pl.duration(days=7))).cast(pl.Int8).alias("_in_7d"),
            (pl.col("t_dat") > (cutoff_expr - pl.duration(days=30))).cast(pl.Int8).alias("_in_30d"),
        ])
    )

    return (
        lf_hist
        .group_by("customer_id")
        .agg([
            pl.len().alias("u_txn_cnt_all"),
            pl.col("article_id").n_unique().alias("u_unique_article_cnt"),
            pl.col("_in_7d").sum().alias("u_txn_cnt_7d"),
            pl.col("_in_30d").sum().alias("u_txn_cnt_30d"),
            pl.col("t_dat").max().alias("u_last_purchase_date"),
            pl.col("price").mean().alias("u_avg_price"),
        ])
        .with_columns(
            (cutoff_expr - pl.col("u_last_purchase_date")).dt.total_days().alias("u_recency_days")
        )
    )


def build_article_features(
    lf_tx: pl.LazyFrame,
    lf_articles: pl.LazyFrame,
    cutoff_date: str,
) -> pl.LazyFrame:
    cutoff_expr = pl.lit(cutoff_date).str.to_date()

    lf_hist = (
        lf_tx
        .filter(pl.col("t_dat") <= cutoff_expr)
        .with_columns([
            (pl.col("t_dat") > (cutoff_expr - pl.duration(days=7))).cast(pl.Int8).alias("_in_7d"),
            (pl.col("t_dat") > (cutoff_expr - pl.duration(days=30))).cast(pl.Int8).alias("_in_30d"),
        ])
    )

    return (
        lf_hist
        .group_by("article_id")
        .agg([
            pl.col("_in_7d").sum().alias("a_pop_7d"),
            pl.col("_in_30d").sum().alias("a_pop_30d"),
            pl.col("price").median().alias("a_price_median"),
            pl.len().alias("a_txn_cnt"),
            pl.col("customer_id").n_unique().alias("a_unique_customer_cnt"),
        ])
        .with_columns(
            ((pl.col("a_txn_cnt") - pl.col("a_unique_customer_cnt")) / pl.col("a_txn_cnt")).alias("a_repeat_rate")
        )
        .join(lf_articles.select(["article_id", "product_group_name"]), on="article_id", how="left")
    )
```

```python
def build_user_article_features(
    lf_tx: pl.LazyFrame,
    lf_articles: pl.LazyFrame,
    cutoff_date: str,
) -> pl.LazyFrame:
    # 为了 8GB 内存稳定，这里使用近180天 + 高频用户子集来演示 user-article 亲和特征样例
    cutoff_expr = pl.lit(cutoff_date).str.to_date()
    start_expr = cutoff_expr - pl.duration(days=180)

    lf_hist = lf_tx.filter((pl.col("t_dat") <= cutoff_expr) & (pl.col("t_dat") >= start_expr))

    lf_top_users = (
        lf_hist
        .group_by("customer_id")
        .agg(pl.len().alias("cnt"))
        .sort("cnt", descending=True)
        .limit(50_000)
        .select("customer_id")
    )

    lf_sub = (
        lf_hist
        .join(lf_top_users, on="customer_id", how="inner")
        .join(lf_articles.select(["article_id", "product_group_name"]), on="article_id", how="left")
    )

    lf_user_pref = (
        lf_sub
        .group_by(["customer_id", "product_group_name"])
        .agg(pl.len().alias("grp_cnt"))
        .with_columns(
            pl.col("grp_cnt").rank("dense", descending=True).over("customer_id").alias("rank_in_user")
        )
        .filter(pl.col("rank_in_user") == 1)
        .group_by("customer_id")
        .agg(pl.col("product_group_name").first().alias("u_top_product_group"))
    )

    return (
        lf_sub
        .group_by(["customer_id", "article_id"])
        .agg([
            pl.len().alias("ua_hist_cnt"),
            pl.col("t_dat").max().alias("ua_last_purchase_date"),
            pl.col("product_group_name").first().alias("ua_product_group"),
        ])
        .join(lf_user_pref, on="customer_id", how="left")
        .with_columns([
            (pl.col("u_top_product_group") == pl.col("ua_product_group")).cast(pl.Int8).alias("ua_same_top_group"),
            (cutoff_expr - pl.col("ua_last_purchase_date")).dt.total_days().alias("ua_recency_days"),
        ])
    )
```

```python
CUTOFF_DATE = "2020-09-15"

lf_user_feat = build_user_features(lf_transactions_clean, CUTOFF_DATE)
lf_article_feat = build_article_features(lf_transactions_clean, lf_articles_clean, CUTOFF_DATE)
lf_user_article_feat = build_user_article_features(lf_transactions_clean, lf_articles_clean, CUTOFF_DATE)

df_user_feat_sample = lf_user_feat.sort("u_txn_cnt_all", descending=True).limit(20).collect(engine="streaming")
df_article_feat_sample = lf_article_feat.sort("a_pop_30d", descending=True).limit(20).collect(engine="streaming")
df_user_article_feat_sample = lf_user_article_feat.sort("ua_hist_cnt", descending=True).limit(20).collect(engine="streaming")

print("df_user_feat_sample")
display(df_user_feat_sample)

print("df_article_feat_sample")
display(df_article_feat_sample)

print("df_user_article_feat_sample")
display(df_user_article_feat_sample)
```

## 6. 测试与验收检查


```python
# 测试1: u_recency_days >= 0
viol_recency = (
    lf_user_feat
    .filter(pl.col("u_recency_days") < 0)
    .select(pl.len().alias("n"))
    .collect(engine="streaming")["n"][0]
)

# 测试2: 7天窗口计数不大于30天窗口计数
viol_window = (
    lf_user_feat
    .filter(pl.col("u_txn_cnt_7d") > pl.col("u_txn_cnt_30d"))
    .select(pl.len().alias("n"))
    .collect(engine="streaming")["n"][0]
)

# 测试3: winsorized 价格边界合理
df_q = lf_transactions_clean.select([
    pl.col("price").quantile(0.01).alias("q01"),
    pl.col("price").quantile(0.99).alias("q99"),
    pl.col("price_winsorized").min().alias("w_min"),
    pl.col("price_winsorized").max().alias("w_max"),
]).collect(engine="streaming")

q01, q99 = float(df_q["q01"][0]), float(df_q["q99"][0])
w_min, w_max = float(df_q["w_min"][0]), float(df_q["w_max"][0])

assert int(viol_recency) == 0, f"发现 u_recency_days < 0: {viol_recency}"
assert int(viol_window) == 0, f"发现 7d > 30d: {viol_window}"
assert w_min >= q01 - 1e-12, "winsorized 最小值越界"
assert w_max <= q99 + 1e-12, "winsorized 最大值越界"

print("全部关键测试通过")
print(f"viol_recency={viol_recency}, viol_window={viol_window}, q01={q01:.6f}, q99={q99:.6f}, w_min={w_min:.6f}, w_max={w_max:.6f}")
```
